```python
def fillNone(keyArr):
  for key in keyArr:
    df[key] = df[key].fillna('None')

def fillMean(keyArr):
  for key in keyArr:
    df[key] = df[key].fillna(df[key].mean())

def fillZero(keyArr):
  for key in keyArr:
    df[key] = df[key].fillna(0)

def fillMode(keyArr):
  for key in keyArr:
    df[key] = df[key].fillna(df[key].mode()[0])
```

```python
df.drop(['Alley', 'PoolQC', 'MiscFeature'], axis=1, inplace=True)
fillNone(['FireplaceQu', 'MasVnrType', 'Fence', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1',
          'BsmtFinType2', 'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond'])
fillMean(['LotFrontage', 'MasVnrArea'])
fillZero(['GarageYrBlt'])
fillMode(['Electrical'])

# Display the count of missing values for all columns
print("\nMissing values count after imputation:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# one hot encoding: MSZoning, Street, LandContour, LotConfig, Neighborhood, Condition1, Condition2 (check if we can combine the 2), BldgType, HouseStyle, RoofStyle, RoofMatl, Exterior1st, Exterior2nd(check for combine for 1 & 2), MasVnrType, Foundation, Heating, Electrical, GarageType, SaleType


# Ordinal Encoding: LotShape, Utilities, LandSlope, ExterQual, ExterCond, BsmtQual, BsmtCond, BsmtExposure, BsmtFinType1, BsmtFinType2 (check for combine), HeatingQC, CentralAir, KitchenQual, Functional, FireplaceQu, GarageFinish, GarageQual, GarageCond, PavedDrive, Fence, SaleCondition
```

```python
import pandas as pd

# Define ordinal mappings based on standard house pricing dataset conventions
# Quality/Condition scales: Ex (Excellent), Gd (Good), TA (Typical), Fa (Fair), Po (Poor), None
qual_map = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'None': 0}

# Mapping for BsmtExposure
exposure_map = {'Gd': 4, 'Av': 3, 'Mn': 2, 'No': 1, 'None': 0}

# Mapping for BsmtFinType
fintype_map = {'GLQ': 6, 'ALQ': 5, 'BLQ': 4, 'Rec': 3, 'LwQ': 2, 'Unf': 1, 'None': 0}

# Mapping for Functional
func_map = {'Typ': 7, 'Min1': 6, 'Min2': 5, 'Mod': 4, 'Maj1': 3, 'Maj2': 2, 'Sev': 1, 'Sal': 0}

# Mapping for GarageFinish
garage_map = {'Fin': 3, 'RFn': 2, 'Unf': 1, 'None': 0}

# Mapping for LotShape
shape_map = {'Reg': 3, 'IR1': 2, 'IR2': 1, 'IR3': 0}

# Mapping for LandSlope
slope_map = {'Gtl': 2, 'Mod': 1, 'Sev': 0}

# Mapping for Utilities
util_map = {'AllPub': 3, 'NoSewr': 2, 'NoSeWa': 1, 'ELO': 0}

# Apply transformations
ordinal_cols = {
    'ExterQual': qual_map, 'ExterCond': qual_map, 'HeatingQC': qual_map,
    'KitchenQual': qual_map, 'FireplaceQu': qual_map, 'GarageQual': qual_map, 'GarageCond': qual_map,
    'BsmtQual': qual_map, 'BsmtCond': qual_map,
    'BsmtExposure': exposure_map,
    'BsmtFinType1': fintype_map, 'BsmtFinType2': fintype_map,
    'Functional': func_map,
    'GarageFinish': garage_map,
    'LotShape': shape_map,
    'LandSlope': slope_map,
    'Utilities': util_map,
    'CentralAir': {'Y': 1, 'N': 0},
    'PavedDrive': {'Y': 2, 'P': 1, 'N': 0}
}

for col, mapping in ordinal_cols.items():
    if col in df.columns:
        df[col] = df[col].map(mapping)

# Check for Fence and SaleCondition separately as they are more categorical/nominal
# but can be ranked roughly by desirability
df['Fence'] = df['Fence'].map({'GdPrv': 4, 'MnPrv': 3, 'GdWo': 2, 'MnWw': 1, 'None': 0})

display(df[list(ordinal_cols.keys()) + ['Fence']].head())
```

```python
# 1. Combine Condition1 and Condition2
conditions = pd.unique(df[['Condition1', 'Condition2']].values.ravel())
for cond in conditions:
    if cond != 'Norm': # Use Norm as the baseline
        df[f'Condition_{cond}'] = ((df['Condition1'] == cond) | (df['Condition2'] == cond)).astype(int)

# 2. Combine Exterior1st and Exterior2nd
exteriors = pd.unique(df[['Exterior1st', 'Exterior2nd']].values.ravel())
for ext in exteriors:
    if pd.notna(ext):
        df[f'Exterior_{ext}'] = ((df['Exterior1st'] == ext) | (df['Exterior2nd'] == ext)).astype(int)

# List of other columns for standard one-hot encoding
oh_cols = [
    'MSZoning', 'Street', 'LandContour', 'LotConfig', 'Neighborhood',
    'BldgType', 'HouseStyle', 'RoofStyle', 'RoofMatl', 'MasVnrType',
    'Foundation', 'Heating', 'Electrical', 'GarageType', 'SaleType'
]

# Apply get_dummies and explicitly convert booleans to integers
df = pd.get_dummies(df, columns=oh_cols, drop_first=True, dtype=int)

# Drop the original overlapping columns that were combined manually
df.drop(['Condition1', 'Condition2', 'Exterior1st', 'Exterior2nd'], axis=1, inplace=True)

print(f"New shape after one-hot encoding: {df.shape}")
# Show all columns in the display
with pd.option_context('display.max_columns', None):
    display(df.head())
```

```python
from scipy.stats import skew

# Identify encoded columns from previous steps
encoded_cols = list(ordinal_cols.keys()) + ['Fence'] + [col for col in df.columns if '_' in col]

# Identify original numerical columns that were NOT encoded
# We exclude 'Id' as it is just an index
original_numeric = [col for col in df.columns if col not in encoded_cols and col != 'Id' and df[col].dtype != 'object']

# Calculate skewness
skewness_values = df[original_numeric].apply(lambda x: skew(x.dropna())).sort_values(ascending=False)

print('Skewness of non-encoded numerical columns:')
display(skewness_values)

# Flag high skewness
high_skew_non_encoded = skewness_values[abs(skewness_values) > 0.75]
print(f'\nFound {len(high_skew_non_encoded)} non-encoded columns with high skewness (>0.75):')
display(high_skew_non_encoded)
```

```python
import numpy as np

# Apply log1p transformation to all skewed features (skew > 0.75)
skewed_features = high_skew_non_encoded.index
df[skewed_features] = np.log1p(df[skewed_features])

print(f"Applied log transformation to {len(skewed_features)} columns.")

# Recalculate skewness to verify improvement
skewness_after = df[skewed_features].apply(lambda x: skew(x)).sort_values(ascending=False)
print("\nSkewness after transformation (selected columns):")
display(skewness_after)
```
